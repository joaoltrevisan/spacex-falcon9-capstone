## Preparation, model tuning, and evaluation
The validated matrix has 90 observations and 80 numeric features; `Class` is 0 for an unsuccessful landing and 1 for a successful landing. The data were split into 72 training and 18 test rows with a stratified 80/20 split and seed 2. Standardization is inside a scikit-learn pipeline for linear models and KNN, so each cross-validation fold fits its scaler using only that fold's training rows. Logistic Regression, SVM, KNN, and Decision Tree were tuned with five-fold stratified GridSearchCV; final metrics below use only the held-out test rows.
